### Ej. 3:

Escribir una función con la siguiente signatura **ordenar_dicc(dicc, orden)**. Recibirá como parámetros un diccionario y una dirección de ordenamiento (ascendente, descendente) y por defecto nos retornara un nuevo diccionario donde sus elementos están ordenados.   

```python
def ordenar_dicc(dicc, orden = False):
    sorted_dicc = sorted(dicc, reverse = orden)
    return sorted_dicc
## Recordar que la función 'sorted' simpre retorna un diccionario "nuevo" (o copia).
```

### Ej. 4:

Escribir una función iterativa que calcule la potenciación de un numero. Recibe como parámetros dos números (naturales) A y B devuelve el valor de A**B. 


```python
def potencia_iterativa(a,b):
    result = 1
    for x in range(b):
        result *= a
    return result
```

### Ej. 5:

Escribir una función recursiva que calcule la potenciación de un numero. Recibe como parámetros dos números (naturales) A y B devuelve el valor de A**B. 

```python
def potencia_recursiva(a, b):
    # caso base
    if b <= 0: 
        return 1
    else:
    # llamada recursiva    
        return a * potencia_recursiva(a, b - 1)   
```

### Ej. 9:


Escribir una función que permita ordenar una Lista. Dicha función recibe 3 (tres) argumentos (lista, orden, copia), en donde, lista es la lista a ordenar, orden indica cómo se ordena la lista, puede ser ascendente o descendente, y copy es Verdadero cuando la función retornara una lista nueva. Por defecto, la función nos devuelve una nueva lista ordenada de forma ascendente.

```python
def orden_lista(lista, orden = False, copia = True):
    if copia:
        lista_new = lista.copy()
        lista_new.sort(reverse = orden)
        return lista_new
    else:
        lista.sort(reverse = orden)
        return lista
```
